```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('placement.csv')
df
```

```python
df.head()
```

```python
plt.scatter(df['cgpa'],df['iq'])
plt.xlabel('cgpa')
plt.ylabel('iq')
plt.title('Placement Stats')

```

```python
X=df.iloc[:,0:1]
Y=df.iloc[:,-1]
```

```python
X
```

```python
Y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=2)
```

```python
from sklearn.linear_model import LinearRegression
# consider a variable to fit the data accordingly
lr=LinearRegression()
lr.fit(X_train,Y_train)
```

```python
# Model is trained.
LinearRegression()
```

```python
X_test
```

```python
Y_test
```

Starting Predictions

```python
#FIRST VALUE PREDICTION
#reshaping the value cause data in 2-d format
lr.predict(X_test.iloc[0].values.reshape(1,2))
```

```python
#predicting the 2nd value
lr.predict(X_test.iloc[1].values.reshape(1,2))
```

Actual True Prediction through Graphcial EDA

```python
plt.scatter(df['cgpa'],df['iq'])
plt.plot(X_train,lr.predict(X_train),color='red')
plt.xlabel('cgpa')
plt.ylabel('iq')
plt.title('Placement Stats')
```
